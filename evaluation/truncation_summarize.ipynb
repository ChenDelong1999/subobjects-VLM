```python
import os, json
import matplotlib.pyplot as plt


tokenizer2color = {
    'Patch': 'Grays',
    'ECIT': 'Blues',
    'Mask2Former': 'Greens',
}

dirs = {
    'Patch': {
        1: '/private/home/delong/workspace/subobjects-VLM/runs/clevr_caption/SmolLM2-135M-Instruct-dinov2_small(768px)/patch/0127-1007-patch_10_per_side_random(1t-768px)',
        5: '/private/home/delong/workspace/subobjects-VLM/runs/clevr_caption/SmolLM2-135M-Instruct-dinov2_small(768px)/patch/0127-1007-patch_10_per_side_random(5t-768px)',
        10: '/private/home/delong/workspace/subobjects-VLM/runs/clevr_caption/SmolLM2-135M-Instruct-dinov2_small(768px)/patch/0125-2133-patch_10_per_side_random(10t-768px)',
        15: '/private/home/delong/workspace/subobjects-VLM/runs/clevr_caption/SmolLM2-135M-Instruct-dinov2_small(768px)/patch/0127-1007-patch_10_per_side_random(15t-768px)',
        20: '/private/home/delong/workspace/subobjects-VLM/runs/clevr_caption/SmolLM2-135M-Instruct-dinov2_small(768px)/patch/0125-2133-patch_10_per_side_random(20t-768px)',
        30: '/private/home/delong/workspace/subobjects-VLM/runs/clevr_caption/SmolLM2-135M-Instruct-dinov2_small(768px)/patch/0125-2133-patch_10_per_side_random(30t-768px)',
        40: '/private/home/delong/workspace/subobjects-VLM/runs/clevr_caption/SmolLM2-135M-Instruct-dinov2_small(768px)/patch/0125-2134-patch_10_per_side_random(40t-768px)',
        50: '/private/home/delong/workspace/subobjects-VLM/runs/clevr_caption/SmolLM2-135M-Instruct-dinov2_small(768px)/patch/0125-2134-patch_10_per_side_random(50t-768px)',
        100: '/private/home/delong/workspace/subobjects-VLM/runs/clevr_caption/SmolLM2-135M-Instruct-dinov2_small(768px)/patch/0125-2134-patch_10_per_side_random(100t-768px)',
    },
    'ECIT': {
        1: '/private/home/delong/workspace/subobjects-VLM/runs/clevr_caption/SmolLM2-135M-Instruct-dinov2_small(768px)/directsam/0127-1008-directsam_tiny_sa1b_2ep@0_1(1t-768px)',
        5: '/private/home/delong/workspace/subobjects-VLM/runs/clevr_caption/SmolLM2-135M-Instruct-dinov2_small(768px)/directsam/0127-1008-directsam_tiny_sa1b_2ep@0_1(5t-768px)',
        10: '/private/home/delong/workspace/subobjects-VLM/runs/clevr_caption/SmolLM2-135M-Instruct-dinov2_small(768px)/directsam/0125-2134-directsam_tiny_sa1b_2ep@0_1(10t-768px)',
        15: '/private/home/delong/workspace/subobjects-VLM/runs/clevr_caption/SmolLM2-135M-Instruct-dinov2_small(768px)/directsam/0127-1008-directsam_tiny_sa1b_2ep@0_1(15t-768px)',
        20: '/private/home/delong/workspace/subobjects-VLM/runs/clevr_caption/SmolLM2-135M-Instruct-dinov2_small(768px)/directsam/0125-2134-directsam_tiny_sa1b_2ep@0_1(20t-768px)',
        30: '/private/home/delong/workspace/subobjects-VLM/runs/clevr_caption/SmolLM2-135M-Instruct-dinov2_small(768px)/directsam/0125-2134-directsam_tiny_sa1b_2ep@0_1(30t-768px)',
        40: '/private/home/delong/workspace/subobjects-VLM/runs/clevr_caption/SmolLM2-135M-Instruct-dinov2_small(768px)/directsam/0125-2134-directsam_tiny_sa1b_2ep@0_1(40t-768px)',
        50: '/private/home/delong/workspace/subobjects-VLM/runs/clevr_caption/SmolLM2-135M-Instruct-dinov2_small(768px)/directsam/0125-2134-directsam_tiny_sa1b_2ep@0_1(50t-768px)',
        100: '/private/home/delong/workspace/subobjects-VLM/runs/clevr_caption/SmolLM2-135M-Instruct-dinov2_small(768px)/directsam/0125-2134-directsam_tiny_sa1b_2ep@0_1(100t-768px)',
    },
    'Mask2Former': {
        1: '/private/home/delong/workspace/subobjects-VLM/runs/clevr_caption/SmolLM2-135M-Instruct-dinov2_small(768px)/panoptic/0127-1008-panoptic_mask2former_tiny(1t-768px)',
        5: '/private/home/delong/workspace/subobjects-VLM/runs/clevr_caption/SmolLM2-135M-Instruct-dinov2_small(768px)/panoptic/0127-1008-panoptic_mask2former_tiny(5t-768px)',
        10: '/private/home/delong/workspace/subobjects-VLM/runs/clevr_caption/SmolLM2-135M-Instruct-dinov2_small(768px)/panoptic/0125-2133-panoptic_mask2former_tiny(10t-768px)',
        15: '/private/home/delong/workspace/subobjects-VLM/runs/clevr_caption/SmolLM2-135M-Instruct-dinov2_small(768px)/panoptic/0127-1008-panoptic_mask2former_tiny(15t-768px)',
        20: '/private/home/delong/workspace/subobjects-VLM/runs/clevr_caption/SmolLM2-135M-Instruct-dinov2_small(768px)/panoptic/0125-2133-panoptic_mask2former_tiny(20t-768px)',
        30: '/private/home/delong/workspace/subobjects-VLM/runs/clevr_caption/SmolLM2-135M-Instruct-dinov2_small(768px)/panoptic/0125-2133-panoptic_mask2former_tiny(30t-768px)',
        40: '/private/home/delong/workspace/subobjects-VLM/runs/clevr_caption/SmolLM2-135M-Instruct-dinov2_small(768px)/panoptic/0125-2133-panoptic_mask2former_tiny(40t-768px)',
        50: '/private/home/delong/workspace/subobjects-VLM/runs/clevr_caption/SmolLM2-135M-Instruct-dinov2_small(768px)/panoptic/0125-2133-panoptic_mask2former_tiny(50t-768px)',
        100: '/private/home/delong/workspace/subobjects-VLM/runs/clevr_caption/SmolLM2-135M-Instruct-dinov2_small(768px)/panoptic/0125-2133-panoptic_mask2former_tiny(100t-768px)',
    },
}

# Function to calculate metric (loss) from JSON
def get_metric(path):
    with open(os.path.join(path, 'runs/checkpoint-4080/trainer_state.json')) as f:
        state = json.load(f)
        total_loss = sum(step['loss'] for step in state['log_history'] if 'loss' in step)
        return total_loss / len(state['log_history'])


results_summary = {}
for num_tokens in dirs['Patch'].keys():
    results_summary[num_tokens] = {}

# Plot visualization
plt.figure(figsize=(8, 4))
for tokenizer, results in dirs.items():
    cmap = plt.get_cmap(tokenizer2color[tokenizer])
    x_values, y_values = [], []
    for i, (num_tokens, path) in enumerate(results.items()):
        print(f'Processing {tokenizer} with {num_tokens} tokens...')
        try:
            loss = get_metric(path)
            x_values.append(num_tokens)
            y_values.append(loss)

            results_summary[num_tokens][tokenizer] = loss
            
        except Exception as e:
            print(f"Error processing {path}: {e}")

    # Plot connected dots
    plt.plot(x_values, y_values, marker='o', label=tokenizer, color=cmap(150))

plt.xlabel('Number of Tokens', fontsize=12)
plt.ylabel('Loss', fontsize=12)
plt.title('Loss vs. Number of Tokens', fontsize=14)
plt.legend(title='Tokenizer', fontsize=10)
plt.grid(True)
plt.tight_layout()
plt.show()
```
